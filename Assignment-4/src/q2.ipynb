```python
from sklearn import svm
import pandas as pd    
import numpy as np
from sklearn.metrics import accuracy_score
```

```python
# Hrushikesh
dtrain = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset2\train.csv")
dtest = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset2\test.csv")
dval = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset2\val.csv")
```

```python
xtrain = dtrain.drop(columns=['label']).to_numpy()
ytrain = dtrain['label'].to_numpy()

xtest = dtest.drop(columns=['label']).to_numpy()
ytest = dtest['label'].to_numpy()

xval = dval.drop(columns=['label']).to_numpy()
yval = dval['label'].to_numpy()
```

# Polynomial Kernel

```python
from sklearn import svm

deg = 3
clf = svm.SVC(kernel='poly', degree=deg, C=1)
clf.fit(xtrain, ytrain)
print(f"Poly kernel, C=1, deg={deg}")

ytest_pred = clf.predict(xtest)
ytrain_pred = clf.predict(xtrain)
yval_pred = clf.predict(xval)

print("Train Accuracy: ",  accuracy_score(ytrain_pred, ytrain))

print("Test Accuracy: ",  accuracy_score(ytest_pred, ytest))

print("Validation accuracy: ", accuracy_score(yval_pred, yval))
```

```python
def plot_decision_regions(clf, X, y, title="Decision Region (Train)"):
    # mesh
    pad = 1.0
    x_min, x_max = X[:, 0].min() - pad, X[:, 0].max() + pad
    y_min, y_max = X[:, 1].min() - pad, X[:, 1].max() + pad
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 400),
        np.linspace(y_min, y_max, 400)
    )

    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

    # decision regions
    plt.contourf(xx, yy, Z, alpha=0.3)

    # training points
    plt.scatter(X[:, 0], X[:, 1], c=y, s=35, edgecolors='k')

    # support vectors (works for SVC)
    if hasattr(clf, "support_vectors_"):
        sv = clf.support_vectors_
        plt.scatter(sv[:, 0], sv[:, 1],
                    s=120, facecolors='none', edgecolors='k', linewidths=1.5,
                    label="Support Vectors")
        plt.legend()

    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.title(title)
    plt.tight_layout()
    plt.show()
```

```python
plot_decision_regions(clf, xtrain, ytrain, title="Decision Region (Train Data)")
```

```python
plot_decision_regions(clf, xtest, ytest, title="Decision Region (Test Data)")
```
