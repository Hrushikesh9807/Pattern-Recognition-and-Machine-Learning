```python
from sklearn import svm
import pandas as pd    
import numpy as np
from sklearn.metrics import accuracy_score
```

```python
# Hrushikesh
xtrain = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset3\train_data.csv").to_numpy()
ytrain = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset3\train_label.csv").to_numpy()
xtest = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset3\test_data.csv").to_numpy()
ytest = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset3\test_label.csv").to_numpy()
xval = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset3\val_data.csv").to_numpy()
yval = pd.read_csv(r"H:\Sem-7\CS5691\Assignment-4\Dataset3\val_label.csv").to_numpy()
```

# Polynomial Kernel

```python
from sklearn import svm

deg = 3
clf = svm.SVC(kernel='poly', degree=deg, C=1)
clf.fit(xtrain, ytrain)
print(f"Poly kernel, C=1, deg={deg}")

ytest_pred = clf.predict(xtest)
ytrain_pred = clf.predict(xtrain)
yval_pred = clf.predict(xval)

print("Train Accuracy: ",  accuracy_score(ytrain_pred, ytrain))

print("Test Accuracy: ",  accuracy_score(ytest_pred, ytest))

print("Validation accuracy: ", accuracy_score(yval_pred, yval))
```

```python
import numpy as np

n_train = len(ytrain)
C = clf.C

# find unique support vectors and their max alpha value across subproblems
alpha = np.abs(clf.dual_coef_).max(axis=0)

bounded = np.sum(alpha >= (C - 1e-6))
unbounded = np.sum((alpha > 0) & (alpha < (C - 1e-6)))

p_bounded = 100 * bounded / n_train
p_unbounded = 100 * unbounded / n_train

print(f"Bounded SVs: {p_bounded:.2f}%")
print(f"Unbounded SVs: {p_unbounded:.2f}%")
```
