```python
import random
import numpy as np

def simulate_expected_sum(k=2, trials=10_0000):
    """
    Simulates rolling a fair 6-sided die until cumulative sum is divisible by k.
    
    Parameters:
        k (int): Divisor to check divisibility.
        trials (int): Number of simulation runs.
    
    Returns:
        float: Estimated expected value of the stopping sum.
    """
    results = []
    
    for _ in range(trials):
        total = 0
        while True:
            roll = random.randint(1, 6)
            total += roll
            if total % k == 0:
                results.append(total)
                break
    
    return np.mean(results)

# Example usage:
print("Expected value (k=2):", simulate_expected_sum(k=2, trials=200000)/3.5)
print("Expected value (k=3):", simulate_expected_sum(k=30, trials=200000)/3.5)
```

```python
import random
import numpy as np

def biased_die_roll():
    """Rolls a biased 6-sided die with P(k) = k/21."""
    sides = [1, 2, 3, 4, 5, 6]
    weights = [1, 2, 3, 4, 5, 6]  # proportional to probability
    return random.choices(sides, weights=weights, k=1)[0]

def simulate_expected_sum_biased(k=2, trials=100_000):
    """
    Simulates rolling a biased 6-sided die until cumulative sum is divisible by k.
    
    Parameters:
        k (int): Divisor to check divisibility.
        trials (int): Number of simulation runs.
    
    Returns:
        float: Estimated expected value of the stopping sum.
    """
    results = []
    
    for _ in range(trials):
        total = 0
        while True:
            roll = biased_die_roll()
            total += roll
            if total % k == 0:
                results.append(total)
                break
    
    return np.mean(results)

# Example usage:
print("Expected value (k=2):", simulate_expected_sum_biased(k=2, trials=200000))
print("Expected value (k=3):", simulate_expected_sum_biased(k=3, trials=200000))
```
